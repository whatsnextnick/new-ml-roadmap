# 01. Tensors & Tensor Operations

Covering tensor creation, indexing, reshaping, CUDA device placement, and linear algebra operations in PyTorch.

```python
import torch
import numpy as np
```

```python
data = [[1,2],[3,4],[5,6]]
x_data = torch.tensor(data)
```

```python
#tensor made from numpy array
np_array = np.array(data)
x_np = torch.from_numpy(np_array)
```

```python
print(x_np)
```

```python
# You can also create tensors from other tensors
# To retain its properties and data type:
x_ones = torch.ones_like(x_data)
print(f'the ones tensor is {x_ones }.')

x_rand = torch.rand_like(x_data, dtype=torch.float)
print(f'the random tensor is {x_rand }.')
```

```python
shape = (3,5) # (columns, rows)
shape_rand =  torch.rand(shape)
shape_ones = torch.ones(shape)
shape_zeros = torch.zeros(shape)
print(f'the random tensor is:\n {shape_rand}, \n the ones tensor is: \n{shape_ones},\n the zeros tensor is: \n{shape_zeros}')
```

```python
tensor = torch.rand(3,4)
tensor_zeros = torch.zeros(3,4)

print(f"shape of the tensor: {tensor.shape}")
print(f"datatype of the tensor: {tensor.dtype}")
print(f"datatype of the 'tensor_zeros' is: {tensor_zeros.dtype}")
print(f"device tensor is stored on: {tensor.device}")
```

```python
print(tensor)
print(tensor_zeros)
```

```python
print(f"1st row of the tensor is {tensor[0]}")
print(f"1st element of the 1st row is {tensor[0][0]}")
print(f"last row of the tensor is {tensor[-1]}")
print(f"last element of the last row is {tensor[-1][-1]}")
middle_i = tensor[shape[0]//2]
# print(f"middle row is {middle_i}")
# # remember that float round down to the nearest whole int
# print(f"middle element of the middle row is {middle_i[len(middle_i)//2]}")
# print(f"second to last element of the middle row is {tensor[shape[0]//2][-2]}")
middle_num_i = tensor.shape[0] // 2
print(f"middle row of the tensor is {tensor[middle_num_i]}")
print(f"middle element of the middle row is {tensor[middle_num_i][len(tensor[middle_num_i])//2]}")
```

```python
tensor_1 = torch.rand(2,2)
tensor_2 = torch.rand(2,2)
print(tensor_1)
print(tensor_2)
```

```python
concat_tensor = torch.cat([tensor_1,tensor_2]) #expects a tuple ([])
concat_tensor_w_dimension0 = torch.cat([tensor_1,tensor_2], dim=0) #the x of shape = (x,y)
concat_tensor_w_dimension1 = torch.cat([tensor_1,tensor_2], dim=1) #the y of shape
concat_tensor_w_dimension_ne2 = torch.cat([tensor_1,tensor_2], dim=-2) #na so it defaults to 0
print(concat_tensor)
print(concat_tensor_w_dimension0)
print(concat_tensor_w_dimension1)
print(concat_tensor_w_dimension_ne2)
```

```python
# Matrix Multiplication
matrix_product = concat_tensor @ concat_tensor_w_dimension0.T #multiplying the matrix by the transposed matrix
print(f"the matrix product is {matrix_product}")
```

```python
#Element-wise multiplication
elewise_product = concat_tensor * concat_tensor_w_dimension0
print(f"the element wise product is {elewise_product}")
```
